```python
# @title 1. Setup — clone + install (chạy ô này trước)
import os, subprocess, sys

REPO_NAME = "Day21-Track3-Finetuning-Lab-2A202601649-LeCongDung"
REPO = f"https://github.com/ledung20053001-dev/{REPO_NAME}"

# 1. Clone nếu chưa có thư mục và chưa ở trong repo
if not os.path.exists(REPO_NAME) and not os.path.basename(os.getcwd()) == REPO_NAME:
    subprocess.run(["git", "clone", "-q", REPO], check=True)

# 2. Chuyển vào thư mục repo nếu hiện tại đang ở ngoài
if os.path.exists(REPO_NAME):
    os.chdir(REPO_NAME)

# 3. Kéo code mới và cấu hình sys.path
subprocess.run(["git", "pull", "-q"], check=False)
if "src" not in sys.path:
    sys.path.insert(0, "src")

# 4. Cài đặt dependencies từ requirements.txt
subprocess.run([sys.executable, "-m", "pip", "install", "-q", "-r", "requirements.txt"],
               check=True)

import torch
print("Directory:", os.getcwd())
print("Commit   :", subprocess.run(["git", "rev-parse", "--short", "HEAD"],
                                  capture_output=True, text=True).stdout.strip())
print("GPU      :", torch.cuda.get_device_name(0) if torch.cuda.is_available()
      else "NONE — Runtime > Change runtime type > T4 GPU")
if torch.cuda.is_available():
    print("VRAM     : %.1f GB" % (torch.cuda.get_device_properties(0).total_memory / 1024**3))
```
